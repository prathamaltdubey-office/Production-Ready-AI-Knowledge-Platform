```python
from agent.agent import run_agent

question = (
    "What is SmartTech Co.'s market share according to the "
    "SmartHome Hub document, and calculate what 35% of the "
    "projected $135.3 billion smart home market would be?"
)

result = run_agent(question)

print("=" * 70)
print("AGENT EXECUTION TRACE")
print("=" * 70)

print("\nUSER QUESTION:")
print(question)

print("\n" + "-" * 70)
print("EXECUTION STEPS")
print("-" * 70)

step = 1

for message in result["messages"]:

    print(f"\nSTEP {step}")
    print(f"Message type: {message.type}")

    if message.type == "human":
        print("Role: USER")
        print("Content:")
        print(message.content)

    elif message.type == "ai":
        print("Role: AGENT")

        if getattr(message, "tool_calls", None):
            print("Tool calls:")

            for tool_call in message.tool_calls:
                print(f"  - {tool_call['name']}")
                print(f"    Arguments: {tool_call['args']}")
        else:
            print("Agent response:")
            print(message.content)

    elif message.type == "tool":
        print("Role: TOOL")
        print(f"Tool name: {message.name}")
        print("Tool result:")
        print(message.content)

    step += 1

print("\n" + "-" * 70)
print("TRACE SUMMARY")
print("-" * 70)

tool_calls = sum(
    len(getattr(message, "tool_calls", []))
    for message in result["messages"]
    if message.type == "ai"
)

print(f"Total messages: {len(result['messages'])}")
print(f"Total tool calls: {tool_calls}")

print("\n" + "=" * 70)
print("EXECUTION TRACE COMPLETED")
print("=" * 70)
```

```python
from agent.agent import run_agent

print("=" * 70)
print("AGENT ERROR RECOVERY DEMO")
print("=" * 70)


questions = [
    "Calculate 100 divided by 0.",
    "Read the file does_not_exist.txt.",
    "Search Wikipedia for an empty query.",
]


for i, question in enumerate(questions, start=1):

    print("\n" + "-" * 70)
    print(f"TEST {i}")
    print("-" * 70)

    print("\nUSER:")
    print(question)

    try:
        result = run_agent(question)

        print("\nAGENT:")

        final_message = result["messages"][-1]

        print(final_message.content)

        print("\nEXECUTION:")

        for message in result["messages"]:

            if message.type == "ai":
                tool_calls = getattr(message, "tool_calls", [])

                for tool_call in tool_calls:
                    print(
                        f"Tool selected: {tool_call['name']}"
                    )

            elif message.type == "tool":
                print(
                    f"Tool result from {message.name}:"
                )
                print(message.content)

    except Exception as e:

        print("\nAGENT ERROR:")
        print(str(e))


print("\n" + "=" * 70)
print("ERROR RECOVERY DEMO COMPLETED")
print("=" * 70)
```

```python
from agent.agent import run_agent
from rag.generation.rag_chain import RAGChain

print("=" * 70)
print("STANDARD RAG VS AGENTIC AI")
print("=" * 70)

rag = RAGChain()

questions = [
    "What are the key features of the SmartHome Hub?",
    "What is SmartTech Co.'s market share?",
    "What is 125 multiplied by 48?",
]

for i, question in enumerate(questions, start=1):

    print("\n" + "=" * 70)
    print(f"QUESTION {i}")
    print("=" * 70)

    print("\nQuestion:")
    print(question)

    # ---------------------------------------------------------
    # STANDARD RAG
    # ---------------------------------------------------------

    print("\n" + "-" * 70)
    print("STANDARD RAG")
    print("-" * 70)

    rag_result = rag.answer(question)

    print("\nAnswer:")
    print(rag_result["answer"])

    print("\nSources:")
    for source in rag_result["sources"]:
        print("-", source)

    # ---------------------------------------------------------
    # AGENTIC AI
    # ---------------------------------------------------------

    print("\n" + "-" * 70)
    print("AGENTIC AI")
    print("-" * 70)

    agent_result = run_agent(question)

    print("\nExecution:")

    for message in agent_result["messages"]:

        if message.type == "ai":

            tool_calls = getattr(message, "tool_calls", [])

            for tool_call in tool_calls:
                print(
                    f"Tool selected: {tool_call['name']}"
                )

        elif message.type == "tool":

            print(
                f"Tool result: {message.name}"
            )

    print("\nFinal Answer:")

    for message in reversed(agent_result["messages"]):

        if message.type == "ai" and message.content:
            print(message.content)
            break


print("\n" + "=" * 70)
print("COMPARISON COMPLETED")
print("=" * 70)
```
